# Ransomware Detection – Exploratory Data Analysis

This notebook walks through:
1. Loading raw system-metrics and file-event logs
2. Extracting features via the pipeline in `src/features/extractor.py`
3. Visualising the resulting feature distributions

```python
import sys
import os
sys.path.insert(0, os.path.join(os.getcwd(), ".."))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.features.extractor import (
    load_system_metrics,
    extract_system_features,
    load_file_events,
    extract_file_features,
)

%matplotlib inline
sns.set_theme(style='whitegrid')
```

## 1. Load raw data

Replace the paths below with the actual CSV files collected from your VM.

```python
SYSTEM_METRICS_CSV = "../data/raw/system_metrics.csv"
FILE_EVENTS_CSV    = "../data/raw/file_events.csv"

sys_df   = load_system_metrics(SYSTEM_METRICS_CSV)
file_df  = load_file_events(FILE_EVENTS_CSV)

print("System metrics:", sys_df.shape)
sys_df.head()
```

## 2. Extract features

```python
sys_feat  = extract_system_features(sys_df)
file_feat = extract_file_features(file_df)

sys_feat.head()
```

## 3. Visualise CPU and memory usage

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=True)

sys_feat.plot(x='timestamp', y='cpu_percent', ax=axes[0], legend=False)
axes[0].set_ylabel('CPU %')

sys_feat.plot(x='timestamp', y='mem_percent', ax=axes[1], color='orange', legend=False)
axes[1].set_ylabel('Memory %')

plt.tight_layout()
plt.savefig('../reports/cpu_mem_usage.png', dpi=150)
plt.show()
```

## 4. File-event counts

```python
event_cols = [c for c in file_feat.columns if c.startswith('file_')]
file_feat[event_cols].sum().plot(kind='bar')
plt.title('Total file-system events')
plt.ylabel('Count')
plt.tight_layout()
plt.savefig('../reports/file_event_counts.png', dpi=150)
plt.show()
```
